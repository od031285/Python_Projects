Find the most popular keywords for the current year and during covid

```python
three_keywords=pd.read_csv(Path("three_keywords.csv"))
three_keywords.head()
```

```python
three_keywords['month']=pd.to_datetime(three_keywords['month'])
three_keywords['month_num'] = three_keywords['month'].dt.month
three_keywords['year'] = three_keywords['month'].dt.year
```

```python
sns.lineplot(
    data=three_keywords,
    x='year',
    y='home_workout_worldwide',
    color='blue',
    label='home_workout',errorbar=None
)

sns.lineplot(
    data=three_keywords,
    x='year',
    y='gym_workout_worldwide',
    color='green',
    label='gym_workout',errorbar=None
)

sns.lineplot(
    data=three_keywords,
    x='year',
    y='home_gym_worldwide',
    color='red',
    label='home_gym',errorbar=None
)

plt.title('Searches for home workout, gym workout and home gym')
plt.legend()
plt.show()
```
